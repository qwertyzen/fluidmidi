```python
import os
from pathlib import Path
import time
from fsynth.lib import *
from fsynth.metronome import Metronome
from fluidmidi.pyside.midi import *
from fluidmidi.fluidseq import *
import numpy as np
```

```python
path = Path('~/Downloads').expanduser()
mfs, sfs = [], []
for item in os.listdir(path):
    if item.lower().endswith(('.sf2', '.sf3')):
        sfs.append(item)
    if item.lower().endswith('.mid'):
        mfs.append(item)

print(mfs)
# exit()
mf = mfs[4]
sf = sfs[-6]
```

```python
settings = Settings()
synth = Synthesizer(settings)
seq = Sequencer(synth)
```

```python
au = AudioDriver(settings, synth)
```

```python
synth.sfload(os.path.join(path, sf))
```

```python
tempo = 120
seq.set_tempo(tempo, 480)
settings.set('synth.gain', 1.5)
synth.pc(0, 5)
synth.cc(0, ControlEffect1._control, 100)
synth.cc(0, ControlEffect2._control, 100)
```

```python
def create_table():
    BPB = 4
    N = 1 * BPB
    idx = np.arange(N)
    ft = FluidMidi(N)
    dur = 480
    ft['mtype'] = Mtype.NON
    ft['channel'] = 0
    ft['time'] = idx * dur
    ft['noteval'] = [60, 64, 67, 72]
    ft['notevel'] = np.where(np.mod(idx, BPB), 60, 90)
    ft['midi_d'] = dur
    return ft.make()
```

```python
tbl = create_table()
tbl.print()
tbl.ndarray
```

```python
metro = Metronome(seq, tempo, 8)
```

```python
flseq = FluidMidiSequence(seq, tbl, 0, 0)
```

```python
flseq.start()
metro.start()
```

```python
metro.toggle_mute()
```

```python
flseq.stop()
metro.stop()
```
